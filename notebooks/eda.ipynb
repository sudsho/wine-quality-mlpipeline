# Wine Quality EDA

Quick look at the UCI Wine Quality dataset before throwing a model at it.

```python
import os
import sys
sys.path.append('..')
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
from src.load_data import load_red, load_white, load_combined
```

```python
red = load_red()
white = load_white()
print(red.shape, white.shape)
red.head()
```

## Quality distribution

Heavy class imbalance: most wines sit at quality 5 or 6. Tails (3 and 8) are tiny.
We will collapse to a binary 'good wine' label using quality >= 7 to make the
problem tractable for a basic random forest.

```python
fig, axes = plt.subplots(1, 2, figsize=(10,3))
red['quality'].value_counts().sort_index().plot.bar(ax=axes[0], title='red')
white['quality'].value_counts().sort_index().plot.bar(ax=axes[1], title='white')
plt.tight_layout()
```

## Feature distributions

```python
red.hist(figsize=(12,8), bins=30)
plt.tight_layout()
```

## Correlation heatmap

```python
import numpy as np
corr = red.corr()
fig, ax = plt.subplots(figsize=(8,6))
cax = ax.matshow(corr, cmap='coolwarm', vmin=-1, vmax=1)
fig.colorbar(cax)
ax.set_xticks(range(len(corr.columns)))
ax.set_yticks(range(len(corr.columns)))
ax.set_xticklabels(corr.columns, rotation=90)
ax.set_yticklabels(corr.columns)
plt.title('red wine feature correlations', pad=20)
```

Alcohol shows the strongest positive correlation with quality, volatile acidity
the strongest negative one. That matches the wine literature.
